## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

```

Research Question:
Which content pages should be reviewed first for refresh or
improvement, based on observable search and engagement signals?

Decision: Which pages does a content reviewer look at first?

Action: Review and refresh the highest-priority pages
(update content, fix CTR issues, improve engagement).
Cost of a wrong recommendation:
- False positive: reviewer wastes time on a page that did not need fixing.
- False negative: a truly declining page gets ignored and loses more traffic.

Why ML can help:
A simple rule (like "declining pages with impressions > 100") misses
complex combinations of signals. A learned model can find patterns
across position, CTR, age, engagement, and trend together.
