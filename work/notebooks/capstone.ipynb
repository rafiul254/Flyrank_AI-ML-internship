## Part 1: Demo Outline (5-Minute Showcase Script)

> **Format:** One question → one method → one chart → one honest result → one recommendation.
> Timing is approximate. Designed to stand alone without slides.

---

### ⏱ 0:00 – 0:45 · The Question

> *"Search content decays. Rankings drop slowly, week over week, and by the time the traffic loss is visible, the opportunity cost is already done. The question I worked on: can a machine learning model predict which pages are on a declining trend — before a human editor would catch it — using nothing but production search signals that any SEO team already has?"*

- Real problem: editorial teams waste capacity on blanket audits or miss decay entirely
- Value: a prioritised refresh queue → editors act on the right pages, earlier

---

### ⏱ 0:45 – 2:15 · The Method

> *"Here's how I approached it."*

1. **Dataset:** 30,000 page-level records aggregated from ~79 million rows of real production impression and click data (FlyRank ML Internship Dataset)
2. **Label:** `is_refresh_opportunity = 1` where `trend_direction == 'down'` — binary classification, ~36% positive rate
3. **Model:** Random Forest (scikit-learn, 100 estimators, class-balanced)
4. **Leakage audit — this is the most important part:**
   - Found `trend_pct` directly encodes the label → removed (Level-1 leakage)
   - Found `impression_momentum` is a near-perfect indirect proxy → removed (Level-2)
   - Without this, AUC looked like 0.97. That's a ghost metric.
5. **Validation:** Three split strategies compared — random, grouped-by-client (primary), time-aware
   - Grouped-by-client: no client appears in both train and test → honest generalisation estimate

---

### ⏱ 2:15 – 3:15 · One Chart

> *"The chart I'd show is **Precision@K** — the rule-based baseline vs the model on the same grouped holdout."*

**→ Figure 3 from the paper (Precision@K bar chart)**

- X-axis: K = 100, 200, 500, 1000
- At K=500: model Precision = **0.52**, baseline Precision = **0.38**
- **Why this chart?** Precision@K is the operationally real metric — editors act on a ranked list, not on 30,000 rows. This chart directly answers: "if I hand this list to an editor, how much of it is useful?"

---

### ⏱ 3:15 – 4:00 · One Honest Result

> *"The headline number is a 37% lift over the rule-based baseline at K=500, on the grouped holdout. But the honest story is what happened before that number."*

- Random split AUC: **0.97** → sounds great, it's a lie (same-client rows in train and test)
- Grouped split AUC (clean features): **0.79** → real, credible, and still useful
- The leakage removal dropped AUC from 0.97 to 0.79 — that delta *is* the lesson
- This is directional signal for triage, not a production-ready prediction engine

---

### ⏱ 4:00 – 5:00 · One Recommendation

> *"If I had to give one recommendation based on this work:"*

**Use the model as a triage filter, not a decision-maker.**

Specifically:
- Run the model weekly on new content snapshots
- Surface the top-N pages where **model and rule-based baseline agree** (consensus = highest confidence)
- Sort those by `impressions_90d` descending — refresh the high-traffic at-risk pages first
- Before any production deployment: retrain on a time-aware split and validate on fresh data

> *"The model doesn't tell you what to write. It tells you where to look."*
